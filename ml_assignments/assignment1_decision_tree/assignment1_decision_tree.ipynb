# Assignment 1: Decision Tree for Student Performance

**Objective:** Design and train a Decision Tree model to predict student performance and compare the effect of `max_depth = 3, 4, 5`.

### Features used
- Study hours
- Attendance percentage
- Previous score
- Assignment completion percentage
- Sleep hours
- Internet access

The same train/test split is used for all three models so the comparison is fair.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

RANDOM_STATE = 42
```

```python
data = [
    (1.0, 62, 48, 55, 5.0, 0, "Needs Improvement"),
    (1.5, 65, 52, 60, 5.5, 0, "Needs Improvement"),
    (2.0, 68, 55, 65, 6.0, 1, "Needs Improvement"),
    (2.5, 70, 58, 68, 6.0, 1, "Needs Improvement"),
    (3.0, 72, 61, 70, 6.5, 1, "Needs Improvement"),
    (3.0, 75, 63, 72, 6.5, 1, "Good"),
    (3.5, 76, 65, 75, 7.0, 1, "Good"),
    (3.5, 78, 67, 78, 7.0, 1, "Good"),
    (4.0, 80, 70, 80, 7.0, 1, "Good"),
    (4.0, 82, 72, 82, 7.5, 1, "Good"),
    (4.5, 84, 74, 85, 7.5, 1, "Good"),
    (4.5, 86, 76, 88, 8.0, 1, "Good"),
    (5.0, 88, 78, 90, 8.0, 1, "Good"),
    (5.0, 90, 80, 92, 8.0, 1, "Good"),
    (5.5, 91, 82, 94, 8.5, 1, "Good"),
    (6.0, 93, 85, 96, 8.5, 1, "Good"),
    (1.0, 60, 45, 50, 5.0, 0, "Needs Improvement"),
    (1.5, 63, 50, 58, 5.5, 1, "Needs Improvement"),
    (2.0, 66, 53, 62, 6.0, 1, "Needs Improvement"),
    (2.5, 69, 57, 66, 6.0, 1, "Needs Improvement"),
    (3.0, 73, 60, 71, 6.5, 1, "Good"),
    (3.5, 77, 64, 76, 7.0, 1, "Good"),
    (4.0, 79, 69, 81, 7.0, 1, "Good"),
    (4.5, 83, 73, 84, 7.5, 1, "Good"),
    (5.0, 87, 77, 89, 8.0, 1, "Good"),
    (5.5, 89, 81, 93, 8.5, 1, "Good"),
    (1.0, 61, 47, 52, 5.0, 0, "Needs Improvement"),
    (2.0, 67, 54, 64, 6.0, 0, "Needs Improvement"),
    (2.5, 71, 59, 69, 6.5, 1, "Needs Improvement"),
    (3.0, 74, 62, 73, 6.5, 1, "Good"),
    (3.5, 79, 66, 77, 7.0, 1, "Good"),
    (4.0, 81, 71, 83, 7.5, 1, "Good"),
    (4.5, 85, 75, 87, 8.0, 1, "Good"),
    (5.0, 88, 79, 91, 8.0, 1, "Good"),
    (5.5, 92, 83, 95, 8.5, 1, "Good"),
    (1.5, 64, 49, 56, 5.5, 0, "Needs Improvement"),
    (2.5, 69, 56, 67, 6.0, 1, "Needs Improvement"),
    (3.0, 76, 64, 74, 7.0, 1, "Good"),
    (3.5, 80, 68, 79, 7.0, 1, "Good"),
    (4.0, 83, 72, 85, 7.5, 1, "Good"),
    (4.5, 86, 76, 88, 8.0, 1, "Good"),
]

columns = [
    "study_hours", "attendance_percent", "previous_score",
    "assignments_completed_percent", "sleep_hours",
    "internet_access", "performance"
]

df = pd.DataFrame(data, columns=columns)
df.head()
```

## 2. Prepare the data

```python
X = df.drop(columns="performance")
y = df["performance"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 3. Train Decision Trees with max_depth = 3, 4 and 5

```python
results = []
models = {}

for depth in [3, 4, 5]:
    model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=RANDOM_STATE
    )
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    results.append({
        "max_depth": depth,
        "accuracy": accuracy_score(y_test, predictions),
        "precision": precision_score(y_test, predictions, pos_label="Good", zero_division=0),
        "recall": recall_score(y_test, predictions, pos_label="Good", zero_division=0),
        "f1_score": f1_score(y_test, predictions, pos_label="Good", zero_division=0)
    })
    models[depth] = model

results_df = pd.DataFrame(results)
results_df.round(3)
```

## 4. Comparative analysis

```python
print("Comparative Analysis:")
print("- max_depth=3: simpler tree and lower complexity, but it may underfit.")
print("- max_depth=4: allows more decision rules and can capture additional patterns.")
print("- max_depth=5: more complex tree; extra complexity can increase overfitting risk.")
print("\nUse the test-set metrics above to compare generalization on this dataset.")
```

```python
plot_df = results_df.melt(
    id_vars="max_depth",
    value_vars=["accuracy", "precision", "recall", "f1_score"],
    var_name="metric",
    value_name="score"
)

plt.figure(figsize=(9, 5))
sns.barplot(data=plot_df, x="max_depth", y="score", hue="metric")
plt.ylim(0, 1.05)
plt.title("Decision Tree Performance for Different max_depth Values")
plt.xlabel("max_depth")
plt.ylabel("Score")
plt.show()
```

## 5. Visualize the Decision Tree

```python
plt.figure(figsize=(18, 9))
plot_tree(
    models[4],
    feature_names=X.columns,
    class_names=models[4].classes_,
    filled=True,
    rounded=True,
    fontsize=8
)
plt.title("Decision Tree - max_depth=4")
plt.show()
```

### Conclusion

Decision Tree classification was successfully implemented for student-performance prediction. Three tree depths (`3`, `4`, and `5`) were tested and compared using accuracy, precision, recall and F1-score. Increasing depth makes the tree more complex; therefore, depth should be selected by considering test performance and the risk of overfitting.
