# Assignment 3: Unsupervised Learning using K-Means

**Dataset:** Iris dataset

**Objective:** Demonstrate unsupervised learning by grouping flowers into clusters without using the species labels during training.

K-Means is used to create 3 clusters. PCA is then used only to visualize the four-dimensional dataset in two dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

RANDOM_STATE = 42
```

## 1. Load the Iris dataset

```python
iris = load_iris()

X = pd.DataFrame(
    iris.data,
    columns=[
        "sepal_length_cm",
        "sepal_width_cm",
        "petal_length_cm",
        "petal_width_cm"
    ]
)

X.head()
```

## 2. Standardize the features

Standardization puts all features on a comparable scale. This is useful for K-Means because K-Means uses distances between data points.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 3. Train the K-Means model

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=RANDOM_STATE,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

X_clustered = X.copy()
X_clustered["cluster"] = clusters

X_clustered.head()
```

## 4. Evaluate the clustering

```python
silhouette = silhouette_score(X_scaled, clusters)

print("Number of clusters:", kmeans.n_clusters)
print(f"Silhouette Score: {silhouette:.4f}")

print("\nCluster sizes:")
print(X_clustered["cluster"].value_counts().sort_index())
```

### Important point

The Iris species labels are **not used to train K-Means**. This makes the example unsupervised. The original labels can only be used after training if we want to inspect how the discovered clusters relate to the known species.

## 5. Use PCA for visualization

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=clusters,
    s=50
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering of Iris Dataset")
plt.colorbar(scatter, label="Cluster")
plt.show()
```

### Conclusion

K-Means successfully demonstrated unsupervised learning by grouping the Iris samples into three clusters without using species labels during training. Standardization was applied before clustering, and PCA was used to visualize the clusters in two dimensions. The silhouette score provides a numerical measure of cluster separation.
