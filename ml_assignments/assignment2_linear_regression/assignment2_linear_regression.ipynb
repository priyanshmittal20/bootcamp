# Assignment 2: Linear Regression in the Electronics Industry

**Application:** Predict rechargeable-battery terminal voltage from discharge current and temperature.

This is a simplified educational electronics example. The dataset represents controlled battery-discharge measurements.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
```

## 1. Create the electronics dataset

```python
rng = np.random.default_rng(RANDOM_STATE)

current = np.array([
    0.5, 0.5, 1.0, 1.0, 1.5, 1.5, 2.0, 2.0,
    2.5, 2.5, 3.0, 3.0, 3.5, 3.5, 4.0, 4.0,
    0.8, 1.8, 2.8, 3.8, 1.2, 2.2, 3.2, 4.2
])

temperature = np.array([
    20, 30, 20, 30, 20, 30, 20, 30,
    20, 30, 20, 30, 20, 30, 20, 30,
    25, 25, 25, 25, 35, 35, 35, 35
])

voltage = (
    12.6
    - 0.42 * current
    - 0.008 * (temperature - 25)
    + rng.normal(0, 0.035, size=len(current))
)

df = pd.DataFrame({
    "discharge_current_A": current,
    "temperature_C": temperature,
    "terminal_voltage_V": voltage
})

df.head()
```

## 2. Split the data

```python
X = df[["discharge_current_A", "temperature_C"]]
y = df["terminal_voltage_V"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=RANDOM_STATE
)
```

## 3. Train the Linear Regression model

```python
model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Model Equation:")
print(
    f"Voltage = {model.intercept_:.4f}"
    f" + ({model.coef_[0]:.4f} × Current)"
    f" + ({model.coef_[1]:.4f} × Temperature)"
)
```

## 4. Evaluate the model

```python
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)

print(f"MAE  : {mae:.4f} V")
print(f"MSE  : {mse:.4f} V²")
print(f"RMSE : {rmse:.4f} V")
print(f"R²   : {r2:.4f}")
```

### Meaning of the metrics

- **MAE:** Average absolute prediction error.
- **MSE:** Average squared prediction error.
- **RMSE:** Error in the same unit as voltage (V).
- **R²:** Shows how much variation in voltage is explained by the model.

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, predictions)
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)
plt.xlabel("Actual Terminal Voltage (V)")
plt.ylabel("Predicted Terminal Voltage (V)")
plt.title("Actual vs Predicted Battery Terminal Voltage")
plt.show()
```

### Conclusion

Linear Regression was used to model the relationship between battery terminal voltage, discharge current and temperature. The model was trained using electronics-related input features and evaluated using MAE, MSE, RMSE and R².
