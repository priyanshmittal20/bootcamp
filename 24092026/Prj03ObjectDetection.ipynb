```python
import cv2
import time
from collections import Counter
from ultralytics import YOLO, settings

# Disable Ultralytics tracking/data sync
settings.update({"sync": False})

# --------------------------------------------------
# 1. Load YOLOv8 Nano model
# --------------------------------------------------
model = YOLO("yolov8n.pt")

# --------------------------------------------------
# 2. Open webcam
# --------------------------------------------------
cap = cv2.VideoCapture(0)

print("AI Object Detection Started!")
print("Press 'q' to quit")
print("Press 's' to save screenshot")

# FPS variables
prev_time = 0

# Vehicle classes
vehicle_classes = ["car", "motorcycle", "bus", "truck"]

while cap.isOpened():

    success, frame = cap.read()

    if not success:
        print("Failed to grab camera frame.")
        break

    # --------------------------------------------------
    # 3. YOLO object detection
    # --------------------------------------------------
    results = model(frame, verbose=False)

    # Get detection result
    result = results[0]

    # Draw bounding boxes
    annotated_frame = result.plot()

    # --------------------------------------------------
    # 4. Count detected objects
    # --------------------------------------------------
    class_names = []

    for box in result.boxes:
        class_id = int(box.cls[0])
        class_name = model.names[class_id]
        class_names.append(class_name)

    # Count each object type
    object_counts = Counter(class_names)

    total_objects = len(class_names)

    # --------------------------------------------------
    # 5. Count vehicles
    # --------------------------------------------------
    vehicle_count = 0

    for obj in class_names:
        if obj in vehicle_classes:
            vehicle_count += 1

    # --------------------------------------------------
    # 6. Calculate FPS
    # --------------------------------------------------
    current_time = time.time()

    fps = 1 / (current_time - prev_time) if prev_time != 0 else 0

    prev_time = current_time

    # --------------------------------------------------
    # 7. Create information panel
    # --------------------------------------------------
    cv2.rectangle(
        annotated_frame,
        (10, 10),
        (330, 180),
        (0, 0, 0),
        -1
    )

    # --------------------------------------------------
    # 8. Display information
    # --------------------------------------------------

    cv2.putText(
        annotated_frame,
        f"Total Objects : {total_objects}",
        (25, 45),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 0),
        2
    )

    cv2.putText(
        annotated_frame,
        f"Vehicles      : {vehicle_count}",
        (25, 75),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 255),
        2
    )

    cv2.putText(
        annotated_frame,
        f"FPS           : {fps:.1f}",
        (25, 105),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 255, 0),
        2
    )

    cv2.putText(
        annotated_frame,
        "Press S = Screenshot",
        (25, 140),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (255, 255, 255),
        1
    )

    cv2.putText(
        annotated_frame,
        "Press Q = Quit",
        (25, 165),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (255, 255, 255),
        1
    )

    # --------------------------------------------------
    # 9. Show live camera
    # --------------------------------------------------
    cv2.imshow(
        "BCA AI Demo - Smart Object Detector",
        annotated_frame
    )

    # --------------------------------------------------
    # 10. Keyboard controls
    # --------------------------------------------------
    key = cv2.waitKey(1) & 0xFF

    # Quit
    if key == ord("q"):
        print("Terminating program...")
        break

    # Screenshot
    if key == ord("s"):
        filename = f"screenshot_{int(time.time())}.jpg"

        cv2.imwrite(filename, annotated_frame)

        print(f"Screenshot saved as {filename}")


# --------------------------------------------------
# 11. Release resources
# --------------------------------------------------
cap.release()

cv2.waitKey(1)
cv2.destroyAllWindows()
cv2.waitKey(1)

print("Camera feed closed successfully.")
```
