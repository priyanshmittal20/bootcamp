```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Provided text
text = """
My favorite hobby is playing cricket. I love playing cricket because it is an interesting and enjoyable sport. I usually play cricket with my friends in the evening or whenever I get free time. Cricket is not only a game for me, but it is also a good way to stay active and fresh. When I play cricket, I feel happy and energetic. I enjoy all parts of the game, including batting, bowling, and fielding. My favorite part is batting because I like hitting the ball and scoring runs for my team.

Playing cricket has taught me many important things. It has taught me the value of teamwork, discipline, patience, and hard work. In cricket, every player has a different role, and everyone has to work together to perform well. Sometimes we win a match, and sometimes we lose, but both experiences teach us something. Winning makes us happy, while losing teaches us to improve our mistakes and try again.

I also enjoy watching professional cricket matches on television and learning new techniques from good players. Cricket helps me take a break from my studies and reduces stress. It also helps me make good friendships with other players. I want to continue playing cricket in the future because it keeps me healthy, active, and confident. Overall, playing cricket is my favorite hobby, and I always enjoy spending my free time on the cricket ground with my friends.
"""
print(text)
```

```python
# Tokenization of words
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
# Frequency distribution of words
word_freq = Counter(filtered_words)
print(word_freq)
```

```python
# Create a WordCloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```
